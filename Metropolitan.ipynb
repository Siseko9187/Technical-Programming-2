# Feature engineering

```python
merged['VoteShare'] = (
    merged['TotalValidVotes'] / 
    merged.groupby(['Year', 'VotingStationName'])['TotalValidVotes'].transform('sum')
)
```

```python
merged['VoteSharePct'] = merged['VoteShare'] * 100
```

```python
merged['TotalVotesCast'] = (
    merged['TotalValidVotes'] + merged['SpoiltVotes']
)
```

```python
merged['TurnoutPct'] = (
    merged['TotalVotesCast'] /
    merged['RegisteredVoters']
) * 100
```

```python
merged['SpoiltVotePct'] = (
    merged['SpoiltVotes'] /
    merged['TotalVotesCast']
) * 100
```

# Machine Learning

```python
# Make sure votes are numeric
merged['TotalValidVotes'] = pd.to_numeric(
    merged['TotalValidVotes'],
    errors='coerce'
).fillna(0)

# Aggregate to Ward + Party + Year
ml_data = (
    merged.groupby(
        ['Year', 'Ward', 'PartyName']
    )
    .agg({
        'TotalValidVotes': 'sum',
        'RegisteredVoters': 'max',
        'SpoiltVotes': 'sum'
    })
    .reset_index()
)
```

```python
ml_data['Winner'] = (
    ml_data.groupby(['Year', 'Ward'])['TotalValidVotes']
    .transform('max')
    == ml_data['TotalValidVotes']
).astype(int)
```

```python
ml_data.head()
```

```python
ml_data = ml_data.sort_values(
    ['Ward', 'PartyName', 'Year']
)
```

```python
ml_data['PreviousVotes'] = (
    ml_data
    .groupby(['Ward', 'PartyName'])['TotalValidVotes']
    .shift(1)
)
```

```python
ml_data['PreviousRegisteredVoters'] = (
    ml_data
    .groupby(['Ward', 'PartyName'])['RegisteredVoters']
    .shift(1)
)
```

```python
ml_data['PreviousSpoiltVotes'] = (
    ml_data
    .groupby(['Ward', 'PartyName'])['SpoiltVotes']
    .shift(1)
)
```

```python
ward_total = (
    ml_data
    .groupby(['Year', 'Ward'])['TotalValidVotes']
    .transform('sum')
)
```

```python
ml_data['VoteShare'] = (
    ml_data['TotalValidVotes'] / ward_total
)
```

```python
ml_data['PreviousVoteShare'] = (
    ml_data
    .groupby(['Ward', 'PartyName'])['VoteShare']
    .shift(1)
)
```

```python
ml_data['VoteChange'] = (
    ml_data['PreviousVotes']
    - ml_data['PreviousVotes'].shift(1)
)
```

```python
ml_data['VoteChange'] = (
    ml_data['PreviousVotes']
    .groupby([ml_data['Ward'], ml_data['PartyName']])
    .diff()
)
```

```python
ml_data = ml_data.dropna(
    subset=[
        'PreviousVotes',
        'PreviousVoteShare'
    ]
)
```

```python
ml_data.shape
```

```python
train = ml_data[
    ml_data['Year'] < 2021
]

test = ml_data[
    ml_data['Year'] == 2021
]
```

```python
features = [
    'PreviousVotes',
    'PreviousVoteShare',
    'PreviousRegisteredVoters',
    'PreviousSpoiltVotes'
]
```

```python
X_train = train[features]
y_train = train['Winner']

X_test = test[features]
y_test = test['Winner']
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight='balanced'
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:20])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    y_pred
)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```
